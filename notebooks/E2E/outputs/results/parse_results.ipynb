```python
from pathlib import Path
import re
import pandas as pd
```

```python
# Scan every immediate sub-folder for a scores.txt
pattern = re.compile(r'(\w+(?:_\w+)?):\s+([0-9.]+)')
records = []

for score_file in Path().glob('*/scores.txt'):
    run = score_file.parent.name
    metrics = {'run': run}
    with score_file.open() as f:
        for line in f:
            m = pattern.match(line.strip())
            if m:
                metrics[m.group(1)] = float(m.group(2))
    records.append(metrics)

# Build a DataFrame and sort by BLEU (change the metric if you prefer)
df = (pd.DataFrame(records)
        .set_index('run')
        .sort_values('BLEU', ascending=False))   # change to 'CIDEr', etc.

display(df)        # use print(df) if you're outside Jupyter
```
