```python
from huggingface_hub import snapshot_download
from pathlib import Path

# Destination directory
model_dir = Path("llama-3.1-8b-instruct")
model_dir.mkdir(exist_ok=True)

# Download (requires HF login if gated)
snapshot_download(
    repo_id="meta-llama/Meta-Llama-3.1-8B-Instruct",
    local_dir=model_dir,
    local_dir_use_symlinks=False,
)
```

```python
from datasets import load_dataset

# Load the dataset (full download, not streaming)
triviaqa = load_dataset("mandarjoshi/trivia_qa", "rc")

# Save to disk
triviaqa.save_to_disk("triviaqa_rc_full")
```

```python
import tarfile

def make_archive(src_dir: str, dest_path: str):
    with tarfile.open(dest_path, "w:gz") as tar:
        tar.add(src_dir, arcname=".")

# Archive model and dataset
make_archive("llama-3.1-8b-instruct", "llama-3.1-8b-instruct.tar.gz")
make_archive("triviaqa_rc_full", "triviaqa_rc_full.tar.gz")
```
